### Tokenizing with code

```python
import tiktoken

encoding = tiktoken.get_encoding("o200k_base")

tokens = encoding.encode("Hi my name is Cytority, I like to torture Brain")
```

# Common Encoding Names

1. o200k_base: Used by models such as gpt-4o and gpt-4o-mini
2. cl100k_base: Used by gpt-4, gpt-4-turbo, gpt-3.5-turbo, and embedding models.
3. p50k_base: Used by Codex models and davinci-002 / davinci-003
4. r50k_base / gpt2: Used by GPT-3 (davinci) and GPT-2 models.

### The Numbers (50k, 100k, 200k)

1. 50k: ~50,257 to 50,284 tokens
2. 100k: 100,277 tokens (100,256 regular + special tokens)
3. 200k: 200,019 tokens (199,998 regular + special tokens)

### The Letters (r, p, cl, o)
OpenAI has not officially published an acronym key in the [tiktoken repository](https://github.com/openai/tiktoken), so these are internal codenames tied to the model generation each tokenizer was built for:
1. r (r50k_base): Regular / Retrained GPT-3 era — The baseline 50k vocabulary used for original GPT-3 models (davinci, curie, babbage, ada), identical in token count to gpt2.
2. p (p50k_base): Programming / Codex era — Built for Codex (code-davinci) and text-davinci-002/003, adding extra whitespace tokens to compress code indentation more efficiently.
3. cl (cl100k_base): Chat / ChatML era — Introduced alongside ChatGPT (gpt-3.5-turbo, gpt-4) and text-embedding-ada-002, doubling the vocabulary and adding special chat-markup tokens.
4. o (o200k_base): Omni (gpt-4o) era — Introduced with GPT-4o ("omni") and used across reasoning models (o1, o3, o4-mini), doubling the vocabulary again for much better multilingual and code compression.|

```python
for token_id in tokens:
  token_text = encoding.decode([token_id])
  print(f"{token_id}={token_text}")
```

## The Illusion of ""MEMORY"
Many of you'll know this already. But for those that don't --- this might be an "WOOWW!!!" moment for you.

```python
import os
from dotenv import load_dotenv
load_dotenv(override = True)
api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
  print("No API key was found - Please check the API_KEY in your env file.")
elif not api_key.startswith("AQ."):
  print("An API key was found, but it doesn't start with AQ.; Please check you're using the right Key")
else:
  print("API key found and looks good so far!")
```

```python
from google import genai
client = genai.Client(api_key=api_key)

response = client.models.generate_content(
model="gemini-3.1-flash-lite",
contents="Tell me about AI Eng"
)
response.text
```

Message to GEMINI is a list of Dicts

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi! I'm Cytority!"}
    ]
```

```python
import requests

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": 'application/json'}

GEMINI_CHAT_URL = "https://generativelanguage.googleapis.com/v1beta/openai/chat/completions"

def ask_gemini(messages):
    payload = {
        "model": "gemini-3.1-flash-lite",
        "messages": messages,
    }
    response = requests.post(GEMINI_CHAT_URL, headers=headers, json=payload)
    response.raise_for_status()
    result = response.json()
    return result["choices"][0]["message"]["content"]
```

```python
ask_gemini(messages)
```
