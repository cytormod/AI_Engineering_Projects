```python
# Connect to OpenAI client library
# A thin wrapper around calls to HTTP endpoints

openai = OpenAI()

# For Gemini, DeepSeek and Groq, we can use the OpenAI python client
# Because Google and DeepSeek have endpoints compatible with OpenAI
# And OpenAI allows you to change the base_url

anthropic_url = "https://api.anthropic.com/v1/"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
deepseek_url = "https://api.deepseek.com"
groq_url = "https://api.groq.com/openai/v1"
grok_url = "https://api.x.ai/v1"
openrouter_url = "https://openrouter.ai/api/v1"
ollama_url = "http://localhost:11434/v1"

anthropic = OpenAI(api_key=anthropic_api_key, base_url=anthropic_url)
gemini = OpenAI(api_key=google_api_key, base_url=gemini_url)
deepseek = OpenAI(api_key=deepseek_api_key, base_url=deepseek_url)
groq = OpenAI(api_key=groq_api_key, base_url=groq_url)
grok = OpenAI(api_key=grok_api_key, base_url=grok_url)
openrouter = OpenAI(base_url=openrouter_url, api_key=openrouter_api_key)
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
```

```python
tell_a_joke = [
  {
    "role": "user",
    "content": "Tell a joke for a student on a journey to becoming an expert in LLM Engineering."
  },
]
```

```python
def call(sdk, model, message, effort):
    response = sdk.chat.completions.create(model=model, messages=message, reasoning_effort=effort)
    display(Markdown(response.choices[0].message.content))
```

```python
call(gemini, "gemini-3.5-flash", tell_a_joke, "low")
```

## A spicy challenge to test the competitive spirit.

```python
dilemma_prompt = """
You and a partner are contestants on a game show. You're each taken to separate rooms and given a choice:
Cooperate: Choose "Share" — if both of you choose this, you each win $1,000.
Defect: Choose "Steal" — if one steals and the other shares, the stealer gets $2,000 and the sharer gets nothing.
If both steal, you both get nothing.
Do you choose to Steal or Share? Pick one.
"""

dilemma = [
    {"role": "user", "content": dilemma_prompt},
]
```

```python
call(ollama, "deepseek-r1:1.5b", dilemma, "low")
```

```python
call(ollama, "deepseek-r1:1.5b", dilemma, "high")
```

```python
call(groq, "openai/gpt-oss-120b", dilemma, 'low')
```

```python
from openai import BadRequestError

try:
    call(grok, "grok-4.3", dilemma, "low")
except BadRequestError as e:
    print(e.response.text)
```

## Gemini and Anthropic Client Library
We're going via the OpenAI Python Client Library, But the other providers have their libraries too

```python
from google import genai

client = genai.Client()

response = client.models.generate_content(
    model="gemini-3.6-flash", contents="Describe the color Blue to someone who's never been able to see in 1 sentence"
)
print(response.text)
```

```python
from anthropic import Anthropic

client = Anthropic()

response = client.messages.create(
    model="claude-sonnet-5-5",
    messages=[{"role": "user", "content": "Describe the color Blue to someone who's never been able to see in 1 sentence"}],
    max_tokens=100 #max_token field is mandatory for Anthropic API calls
)
print(response.content[0].text)
```

## Lightweight LiteLLM

```python
from litellm import completion
response = completion(model="gemini/gemini-4-argon", messages=tell_a_joke)
reply = response.choices[0].message.content
display(Markdown(reply))
```

```python
print(f"Input tokens: {response.usage.prompt_tokens}")
print(f"Output tokens: {response.usage.completion_tokens}")
print(f"Total tokens: {response.usage.total_tokens}")
print(f"Total cost: {response._hidden_params["response_cost"]*100:.4f} cents")
```
